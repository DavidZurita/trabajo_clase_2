```python
import requests
import pprint
import json
import pandas as pd
```

# Consumo de la Api NVD 

```python
url='https://services.nvd.nist.gov/rest/json/cves/2.0'
# url='https://services.nvd.nist.gov/rest/json/cves/2.0?cveId=CVE-2025-21397'
response = requests.get(url)
if response.status_code == 200:
    response = response.json()
    pprint.pprint(response["vulnerabilities"])
    
    #Dentro se encuentra el json de la peticion que esta en la variable response
    # Este es el comando para guardar el json en un archivo

    # with open('cve.json', 'w', encoding='utf-8') as file:
    #     json.dump(response, file, ensure_ascii=False, indent=4)
    
else:
    print("Error en la petición")
```

# Tratamiento de datos de la API NVD

```python
#Funcion obtener los 10 primero elementos de la lista
def obtener():
    resultado=[]
    i=0
    for item in response["vulnerabilities"]:
        resultado.append(item)
        i=i+1
        if i==10:
            break
    return resultado
```

```python
objetos = obtener()
for resultado in objetos:
    print(resultado["cve"])
```

# Obtener si existen valores null o None.

```python
# Crear una DataFrame con los atributos de la lista
df = pd.DataFrame([item['cve'] for item in objetos])
print(df.isnull().sum())
```

# Tipo de Datos de los atributos de la lista 

```python
tipos_datos=df.dtypes
print(tipos_datos)
```

# Conexion de la base de datos

```python
import psycopg2
from sqlalchemy import create_engine
```

```python
# Normalizar el DataFrame
df_normalized = pd.json_normalize(df.to_dict(orient='records'))

# Convertir el dictionary columns a JSON strings
dict_columns = ['descriptions', 'weaknesses', 'configurations', 'references', 'metrics.cvssMetricV2', 'metrics.cvssMetricV31']
for col in dict_columns:
    if col in df_normalized.columns:
        df_normalized[col] = df_normalized[col].apply(json.dumps)
```

```python
# Los 5 columnas de la tabla
df_normalized_columns = df_normalized.iloc[:11, :5]
print(df_normalized_columns)
```

```python
# Creando con el database
def create_database():
    try:
        conn = psycopg2.connect(
            user="postgres",
            password="postgres",
            host="localhost",
            port="5432",
        )
        conn.autocommit = True
        cursor = conn.cursor()

        cursor.execute("SELECT datname FROM pg_database WHERE datname='cve';")
        if not cursor.fetchone():
            cursor.execute("CREATE DATABASE cve;")
            print("Base de datos 'cve' creada correctamente.")
        else:
            print("La base de datos 'CVE' ya existe.")

        cursor.close()
        conn.close()
    except (Exception, psycopg2.Error) as error:
        print("Error while connecting to PostgreSQL", error)
```

```python
def create_tables_dataframe():
    engine = create_engine('postgresql+psycopg2://postgres:tpostgres@localhost:5432/cve')
    df_normalized_columns.to_sql('cve', engine, if_exists='replace', index=False)
```

```python
# Crear la base de datos
create_database()
# Crear las tablas con el dataframe normalizado
create_tables_dataframe()
```
